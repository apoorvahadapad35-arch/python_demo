## What is groupby()?

## What is merge()?
## merge() is used to combine two DataFrames based on one or more common columns.

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Name": ["Apoorva", "Rahul", "Priya"]
})

marks = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Marks": [90, 85, 95]
})

result = pd.merge(students, marks, on="Student_ID")

print(result)
```

## What does the on parameter specify?
## It specifies the common column used to join the DataFrames.

## What is concat()?
## concat() is used to combine two or more DataFrames either row-wise or column-wise.

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Apoorva", "Rahul"],
    "Marks": [90, 85]
})

df2 = pd.DataFrame({
    "Name": ["Priya", "Ravi"],
    "Marks": [95, 80]
})

result = pd.concat([df1, df2])

print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Apoorva", "Rahul"],
    "Marks": [90, 85]
})

df2 = pd.DataFrame({
    "Name": ["Priya", "Ravi"],
    "Marks": [95, 80]
})
result = pd.concat([df1, df2], axis=1)

print(result)
```

## A line plot is used to show the relationship or trend between two sets of data points by connecting them with straight lines.

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
hours = [2, 4, 3, 5, 6]

plt.plot(days, hours)

plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
hours = [2, 4, 3, 5, 6]

plt.plot(days, hours)

plt.title("Study Hours")

plt.xlabel("Days")
plt.ylabel("Hours")

plt.show()
```

## What is a Bar Chart?
## A bar chart is used to compare values among different categories using rectangular bars.

```python
import matplotlib.pyplot as plt

students = ["Apoorva", "Rahul", "Priya", "Ravi"]
marks = [90, 85, 95, 80]

plt.bar(students, marks)

plt.show()
```
