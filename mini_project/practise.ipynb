```python
l=[1,24,45,56,78,]
total=0

for num in l:
    print(total)
    total=total+num

print(total)

```

## Doubling numbers:



```python
l=[1,24,45,56,78,]
dl=[]

for num in l:
    dl.append(num*2)
    print(dl)
```

## Looping through dictionaries


```python
student_marks={"appu":89,"amuu":67,"anu":80}

for student in student_marks.values():
    print(student)
```

```python
tudent_marks={"appu":89,"amuu":67,"anu":80}

for student in student_marks.keys():
    print(student)
```

```python
tudent_marks={"appu":89,"amuu":67,"anu":80}

for student in student_marks.items():
    print(student)
```
